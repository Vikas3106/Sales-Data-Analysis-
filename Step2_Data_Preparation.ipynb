# Sales Data Analysis
## Step 2 – Dataset Selection & Data Preparation

This notebook performs dataset inspection, missing-value handling, duplicate removal, data-type conversion, and preparation of the cleaned dataset.

```python
import pandas as pd
import glob

# Automatically find the CSV file in the project folder
files = glob.glob('*.csv')
print('CSV files found:', files)
df = pd.read_csv(files[0])
df.head()
```

```python
# Data inspection
print('Rows and Columns:', df.shape)
print('\nColumn Names:')
print(df.columns.tolist())
print('\nData Types:')
print(df.dtypes)
print('\nMissing Values:')
print(df.isnull().sum())
print('\nDuplicate Rows:', df.duplicated().sum())
```

```python
# Data cleaning
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
for col in ['Quantity', 'Unit_Price', 'Sales', 'Profit']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Remove duplicate records
df = df.drop_duplicates()

# Handle missing Profit values using median
df['Profit'] = df['Profit'].fillna(df['Profit'].median())

# Remove incomplete/invalid records
df = df.dropna(subset=['Date','Product','Category','Region','Quantity','Unit_Price','Sales','Profit'])
df = df[(df['Quantity'] > 0) & (df['Sales'] >= 0) & (df['Profit'] >= 0)]

print('Cleaned dataset shape:', df.shape)
print('\nMissing values after cleaning:')
print(df.isnull().sum())
print('\nDuplicates after cleaning:', df.duplicated().sum())
```

```python
# Descriptive statistics
df.describe()
```

```python
# Save cleaned dataset
df.to_csv('Cleaned_Sales_Data.csv', index=False)
print('Cleaned_Sales_Data.csv created successfully.')
```
